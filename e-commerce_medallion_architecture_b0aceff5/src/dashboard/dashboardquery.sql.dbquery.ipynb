```sql
SELECT
  product_name,
  category,
  total_revenue,
  total_orders,
  avg_order_value
FROM
  workspace.dev.gold_sales_by_product
ORDER BY
  total_revenue DESC
LIMIT 10;
SELECT
  CASE
    WHEN total_revenue < 1000 THEN '0-1K'
    WHEN total_revenue < 2000 THEN '1K-2K'
    WHEN total_revenue < 5000 THEN '2K-5K'
    WHEN total_revenue < 10000 THEN '5K-10K'
    WHEN total_revenue < 20000 THEN '10K-20K'
    ELSE '20K+'
  END AS revenue_range,
  COUNT(*) AS customer_count
FROM
  workspace.dev.gold_revenue_by_customer
GROUP BY
  CASE
    WHEN total_revenue < 1000 THEN '0-1K'
    WHEN total_revenue < 2000 THEN '1K-2K'
    WHEN total_revenue < 5000 THEN '2K-5K'
    WHEN total_revenue < 10000 THEN '5K-10K'
    WHEN total_revenue < 20000 THEN '10K-20K'
    ELSE '20K+'
  END
ORDER BY
  CASE revenue_range
    WHEN '0-1K' THEN 1
    WHEN '1K-2K' THEN 2
    WHEN '2K-5K' THEN 3
    WHEN '5K-10K' THEN 4
    WHEN '10K-20K' THEN 5
    ELSE 6
  END;
  SELECT
  segment_type,
  customer_count,
  total_revenue,
  avg_revenue
FROM
  workspace.dev.gold_customer_segmentation
ORDER BY
  customer_count DESC;
  SELECT DISTINCT
  customer_segment
FROM
  workspace.dev.gold_revenue_by_customer
ORDER BY
  customer_segment;

  SELECT
  (
    SELECT
      SUM(total_revenue)
    FROM
      workspace.dev.gold_sales_by_product
  ) AS total_revenue,
  (
    SELECT
      SUM(customer_count)
    FROM
      workspace.dev.gold_customer_segmentation
  ) AS total_customers,
  (
    SELECT
      COUNT(DISTINCT product_id)
    FROM
      workspace.dev.gold_sales_by_product
  ) AS total_products
```
