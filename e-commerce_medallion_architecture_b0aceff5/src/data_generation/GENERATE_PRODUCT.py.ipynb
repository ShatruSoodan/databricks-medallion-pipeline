```python
import pandas as pd
import numpy as np
import random

np.random.seed(42)
random.seed(42)

print("=" * 80)
print("📊 STEP 2: Generating products.csv (500 rows)...")
print("=" * 80)

num_products = 500
product_ids = list(range(1, num_products + 1))

# Product categories
categories = ['Electronics', 'Clothing', 'Home & Garden', 'Sports & Outdoors', 'Toys & Games',
              'Books', 'Beauty & Personal Care', 'Automotive', 'Food & Beverage', 'Office Supplies']

# Generate realistic product names
product_name_templates = [
    "Professional {adjective} {item}",
    "{adjective} {item} Pro",
    "Ultra {adjective} {item}",
    "Premium {adjective} {item}",
    "{adjective} {item} Plus"
]

adjectives = ['Smart', 'Wireless', 'Portable', 'Digital', 'Advanced', 'Premium', 'Ergonomic',
              'Compact', 'Durable', 'Lightweight', 'Heavy-Duty', 'High-Performance']

items_by_category = {
    'Electronics': ['Mouse', 'Keyboard', 'Monitor', 'Webcam', 'Headphones', 'Speaker', 'Charger', 'Cable'],
    'Clothing': ['T-Shirt', 'Jeans', 'Jacket', 'Shoes', 'Sweater', 'Dress', 'Shorts', 'Socks'],
    'Home & Garden': ['Lamp', 'Vase', 'Cushion', 'Rug', 'Curtains', 'Plant Pot', 'Mirror', 'Clock'],
    'Sports & Outdoors': ['Basketball', 'Yoga Mat', 'Dumbbell', 'Tent', 'Backpack', 'Water Bottle'],
    'Toys & Games': ['Board Game', 'Puzzle', 'Action Figure', 'Doll', 'Building Blocks'],
    'Books': ['Novel', 'Textbook', 'Cookbook', 'Biography', 'Self-Help Book'],
    'Beauty & Personal Care': ['Shampoo', 'Lotion', 'Face Cream', 'Perfume', 'Soap'],
    'Automotive': ['Car Charger', 'Phone Mount', 'Dash Cam', 'Air Freshener'],
    'Food & Beverage': ['Coffee', 'Tea', 'Snack Pack', 'Energy Drink', 'Protein Bar'],
    'Office Supplies': ['Notebook', 'Pen Set', 'Desk Organizer', 'Stapler', 'File Folder']
}

product_names = []
product_categories = []
prices = []
costs = []

for i in range(num_products):
    category = random.choice(categories)
    item = random.choice(items_by_category.get(category, ['Item']))
    adjective = random.choice(adjectives)
    template = random.choice(product_name_templates)
    
    name = template.format(adjective=adjective, item=item)
    product_names.append(name)
    product_categories.append(category)
    
    # Generate price (with cost being 40-70% of price)
    price = round(random.uniform(10, 500), 2)
    cost = round(price * random.uniform(0.4, 0.7), 2)
    prices.append(price)
    costs.append(cost)

# Stock quantities
stock_quantities = [random.randint(0, 1000) for _ in range(num_products)]
reorder_levels = [int(sq * 0.2) for sq in stock_quantities]  # 20% of stock

products_df = pd.DataFrame({
    'product_id': product_ids,
    'product_name': product_names,
    'category': product_categories,
    'price': prices,
    'cost': costs,
    'stock_quantity': stock_quantities,
    'reorder_level': reorder_levels
})

# Save to Volume
product_path = '/Volumes/workspace/dev/medallion_data/products.csv'
products_df.to_csv(product_path, index=False)

print(f"\n✅ Products.csv generated:")
print(f"   Total rows: {len(products_df):,}")
print(f"   Categories: {len(products_df['category'].unique())}")
print(f"   Avg price: ${products_df['price'].mean():.2f}")
print(f"   Saved to: {product_path}")

print("\nSample products:")
print(products_df.head(10).to_string(index=False))

# Store valid product IDs for orders
valid_product_ids = products_df['product_id'].tolist()

print(f"\nProducts generation complete! ✅")
```
